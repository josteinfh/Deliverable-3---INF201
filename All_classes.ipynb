Character class with warrior and mage subclasses

Attributes for charachter class:
name, hp_max, damage (default damage without weapon), heal_up (default healing amount), weapon

Methods for character class 

attack(): Takes in target as argument. States that if the character has a weapon, than the target should take the damage that is associated with that weapon through self.weapon.roll_damage(). Else default damage (10)

take_damage(): Takes amount as argument, intended to be supplied through the attack method. slef.hp reduces by amount and sets new value to the largest value of self_hp and 0 (no negative hp.)

heal(): Heals the charachter by a default value of 5 if no other amount is specified. Sets self._hp and self.hp_max to the min of these two (so hp cannot go above max). 


Attributes for warrior subclass:
The same as character (inheritance), with the addition of armor 

Methods for warrior subclass:
Modified take_damage(): Calculates a reduced damage due to the warriors armor. At least 1 damage taken anyway. 


Attributes for mage subclass:
The same as character, with the addition of mana and mana_max 

Methods for mage subclass:
Modified attack(): If the mage has 10 or more mana, double damage is dealt at the cost of 10 mana. If not, 5 mana is regenerated up to a max of 30 mana. 





```python
class Character:

    """Represents a character in the RPG"""
    
    def __init__(self, name: str, hp_max: int = 100, damage: int = 10, heal_up: int = 5, weapon: Weapon | None = None) -> None:
        """
        Initialise the Character
        Args:
            name (str): The name of the character.
            hp_max (int): Max hit points (default = 100)
            damage (int): Damage dealt when character attacks (default = 10)
            heal_up (int): Default healing amount (default = 5)
        """
        self.name = name
        self._hp = hp_max  # Note: _hp indicates private attribute
        self.hp_max = hp_max
        self.damage = damage
        self.heal_up = heal_up
        self.weapon = weapon

    @property
    def is_alive(self) -> bool:
        """Check if the character is still alive"""
        return self._hp > 0

    def attack(self, target) -> None:
        """Attack and deal damage to target character"""
        if self.weapon is not None:
            weapon_damage = self.weapon.roll_damage()
            target.take_damage(weapon_damage)
        else:
            target.take_damage(self.damage)

    def take_damage(self, amount: int) -> None:
        """Character takes damage"""
        self._hp -= amount
        self._hp = max(self._hp, 0) # HP cannot drop below 0

    def heal(self, amount: int = None) -> None:
        """Heal the character"""
        heal_amount = amount if amount is not None else self.heal_up
        self._hp += heal_amount
        self._hp = min(self._hp, self.hp_max)  # HP cannot exceed max HP

    def introduce(self) -> None:
        """Creates string that introduces character"""
        print(f"Hi, my name is {self.name}!")

    def __repr__(self) -> str:
        """Representation of character object"""
        return f"<Character name={self.name} hp={self.hp} damage={self.damage}>"

    def __str__(self) -> str:
        """String representation of character object"""
        return f"Character: {self.name} with {self.hp} HP and {self.damage} damage"

    @property
    def hp(self) -> int:
        """Public read-only property to inspect HP without allowing direct, unsafe modification"""
        return self._hp



class Warrior(Character):
    def __init__(self, name, hp_max = 100, damage = 10, heal_up = 5, armor: int = 5, weapon = None):
        super().__init__(name, hp_max, damage, heal_up, weapon)
        self.armor = armor

    def take_damage(self, amount: int) -> None:
        reduced_damage = max(amount-self.armor, 1)
        super().take_damage(reduced_damage)


class Mage(Character):
    def __init__(self, name, hp_max = 100, damage = 10, heal_up = 5, weapon = None, mana: int = 30, mana_max: int =30):
        super().__init__(name, hp_max, damage, heal_up, weapon)
        self.mana = mana 
        self.mana_max = mana_max

    def attack(self, target):
        if self.mana >= 10:
            self.mana -= 10
            spell_damage = self.damage*2
            target.take_damage(spell_damage)
        else:
            self.mana = min(self.mana += 5, mana_max)
            mana = max(mana, 30)
            super().attack(target)

```

Die class

Attributes: Sides (constrained to minimum 2 sides)

Methods: 
roll(): Generates a random integer in range of the number of sides the die has. 

```python
import random

class Dice:

    def __init__(self, sides: int = 6) -> None:
        """Initialise the dice 
            Args: 
            Number of sides 
        """
        if sides < 2: 
            raise ValueError('Die must have at least 2 sides')
        self.sides = sides 


    def roll(self) -> int:
        """ Roll method for rolling dice"""
        dice_roll = random.randint(1, self.sides)
        return dice_roll

    def __repr__(self) -> str:
        return f"Dice(sides = {self.sides})"

    def __call__(self) -> int:
        """ Method allowing individual objects to be called like functions"""
        return self.roll()

```

Weapon class

```python


class Weapon:
    def __init__(self, name: str, damage_dice: Dice):
        """Intialising the weapon class
        2 attributes: name and damage_dice"""

        self.name = name 
        self.damage_dice = damage_dice


    def roll_damage(self):
        return self.damage_dice.roll()
```

Battle class

```python
class Battle():
    """Create a class for battles between heros and enemies"""
    def __init__(self, hero: Character, enemy: Character):
        self.hero = hero 
        self.enemy = enemy 

    def play_round(self) -> None:
        """Method that implements combat logic"""
        self.hero.attack(self.enemy)
        if self.enemy.is_alive:
            self.enemy.attack(self.hero)


    def simulate_battle(self) -> None:
        """Method that simulates a complete battle"""
        while self.hero.is_alive and self.enemy.is_alive:
            self.play_round() 
```
